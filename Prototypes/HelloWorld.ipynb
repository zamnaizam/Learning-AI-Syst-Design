```python
""""IMPORTS"""

from typing import Dict, TypedDict
from langgraph.graph import StateGraph 
```

```python
"""SINGLE NODE"""

class AgentState(TypedDict):
    msg : str

def greeting_node(state: AgentState) -> AgentState:
    """Return a greeting message."""

    state['msg'] = "Hello " + state["msg"] + "! How can I assist you today?"
    return state

graph = StateGraph(AgentState)
graph.add_node("greeter", greeting_node)
graph.set_entry_point("greeter")
graph.set_finish_point("greeter")

app= graph.compile()
result = app.invoke({"msg": "Alice"})

print(result["msg"])  # Output: {'msg': 'Hello Alice! How can I assist you today?'}
```

```python
"""SINGLE NODE EXERCISE"""

class AgentState(TypedDict):
    msg : str
    age : int

def compliment_node(state:AgentState) -> AgentState:
    """Return a compliment message."""
    state['msg'] = "You are looking great today! By the way "+ state["msg"] +", you are " + str(state["age"]) + " years old."
    return state

graph = StateGraph(AgentState)
graph.add_node("compliment", compliment_node)
graph.set_entry_point("compliment")
graph.set_finish_point("compliment")

app = graph.compile()
result = app.invoke({"msg": "Alice", "age": 30})

print(result["msg"])  # Output: {'msg': 'You are looking great today! By the way, you are 30 years old.'}
```
